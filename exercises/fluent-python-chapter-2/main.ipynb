# Comparação de uma lista com 3 floats e uma tupla com 3 floats

```python
test_list = [1.5, 1.7, 1.9]
test_tuple = (1.5, 1.7, 1.9)

print(f"Test float in a list: {sys.getsizeof(test_list)}")
print(f"Test float in a tuple: {sys.getsizeof(test_tuple)}")
```
