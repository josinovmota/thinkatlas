# Comparação de uma lista com 3 floats e uma tupla com 3 floats

Como pode ser observado, uma `lista de floats` tem 88 bytes, enquanto uma `tupla de floats` tem 72 bytes

Isso pode parecer uma diferença minúscula quando estamos trabalhando apenas com 3 tuplas, mas dentro de um ambiente de desenvolvimento, com bilhões de dados, pode fazer uma grande diferença

Algo interessante a ser abordado também é que o livro foca muito sobre não deixar resquícios de dados ao longo do código. Ou seja, você criar uma lista para utilizá-la apenas uma vez e deixá-la criada é um mal hábito

```python
test_list = [1.5, 1.7, 1.9]
test_tuple = (1.5, 1.7, 1.9)

print(f"Test float in a list: {sys.getsizeof(test_list)}")
print(f"Test float in a tuple: {sys.getsizeof(test_tuple)}")
```

# Uma tupla, apesar de ser imutável, pode conter uma estrutura de dados mutável e se mudarmos a estrutura mutável, a tupla muda. Segue o exemplo:

Aqui podemos ver que no primeiro print, a tupla `tupla1` é igual a `tupla2`. Já no segundo print, elas são diferentes. Isso contrasta um pouco com a ideia da imutabilidade de uma tupla. Por mais que elas sejam imutáveis, se você tiver um tipo de dado mutável dentro dela e esse tipo de dado mudar sua estrutura, a tupla também muda

> O livro adverte que tuplas com items mutáveis tendem a ser uma causa de bugs

```python
tupla1 = (1, "gato", [1, 2])
tupla2 = (1, "gato", [1, 2])

print(tupla1 == tupla2)

tupla1[-1].append(3)

print(tupla1 == tupla2)
```

Aqui criamos uma simples função que nos retorna se uma `tupla` pode ou não ser alterada. Na `tupla1` temos uma lista dentro da tupla, na `tupla2` temos uma tupla dentro de uma tupla

```python
def fixed(o):
    try:
        hash(o)
    except TypeError:
        return False
    return True


tupla1 = (1, "gato", [1, 2])
tupla2 = (1, "gato", (1, 2))

print(fixed(tupla1))
print(fixed(tupla2))
```
