# FrenchDeck Example

```python
import collections

# Pra cada símbolo, existem 13 cartas
# Símbolo = naipe

carta = collections.namedtuple("Carta", ["rank", "naipe"])


class Deck:
    rank = [number for number in range(1, 11)] + list("JQK")
    naipe = ("Paus", "Ouros", "Copas", "Espadas")

    def __init__(self):
        self._cards = [carta(rank, naipe) for rank in self.rank for naipe in self.naipe]

    def __len__(self):
        return len(self._cards)

    def __getitem__(self, position):
        return self._cards[position]


baralho_1 = Deck()

print(len(baralho_1))
print(baralho_1[15])
```

# Vector Example

```python
import math


class Vector:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Vector(x={self.x!r}, y={self.y!r})"

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

    def __abs__(self):
        return math.hypot(self.x, self.y)

    def __bool__(self):
        return bool(abs(self))

    def __add__(self, other):
        x = self.x + other.x
        y = self.y + other.y
        return Vector(x, y)

    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)


vec = Vector(3, 4)
vec2 = Vector(9, 10)
vec3 = Vector(0, 0)

# abs
print(abs(vec))
print(abs(vec2))

# add
print(vec + vec2)

# mul
print(vec * 3)

# bool - Fala sobre a magnitude do vetor, ou seja, um vetor(0, 0) é False. Desde que exista um abs(vector) > 0, vira True
print(bool(vec))
print(bool(vec3))

# __repr__ e __str__
print("\r")
print(vec)
vec
```
