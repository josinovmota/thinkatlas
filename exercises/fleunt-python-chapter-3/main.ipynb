# Hashable vs Unhashable

O livro fala sobre tomarmos cuidado ao dizer que uma tupla é `Imutável`. De certa forma ela é, mas as estruturas dentro dela podem ser mutáveis e isso tornar ela mutável. Aqui eu vou fazer um exemplo:

No exemplo a `tupla1` é diferente da `tupla2`. Isso ocorre pois dentro da `tupla2` temos uma estrutura mutável, ou seja, caso essa estrutura mude, o hash da tupla também muda

Vale também notar que `==` também invoca um special method, que é o `__eq__`

```python
tupla1 = (1, 2)
tupla2 = (1, [2])

print(tupla1 == tupla2)
```
